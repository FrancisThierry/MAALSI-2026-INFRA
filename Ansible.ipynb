Création du playbook

```python
%%writefile playbook.yml
---
- name: Test d'exécution locale
  hosts: local
  gather_facts: no
  tasks:
    - name: Afficher un message de test
      ansible.builtin.debug:
        msg: "Ansible fonctionne en local sur Google Colab !"
```

Test du playbook

```python
!ansible-playbook -i inventory.ini playbook.yml
```

Test de gemini

```python
from google.colab import userdata
from google import genai

# 1. Récupération sécurisée de la clé API
api_key = userdata.get('GEMINI_AI_KEY')

# 2. Initialisation du client GenAI
client = genai.Client(api_key=api_key)

# 3. Envoi d'un prompt simple
response = client.models.generate_content(
    model="models/gemini-3.8-flash",
    contents="Explique en deux phrases le principe de l'idempotence dans Ansible."
)

print(response.text)
```
